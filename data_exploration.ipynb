Insights: 

Our initial analysis of the 500 satellite images reveals a total of 959 annotations. On average, each image contains 1.95 fire zones, with some images containing up to 10 distinct burned areas. This confirms that the model must be capable of multi-object detection within a single frame.

The Labeling Paradox:
While the dataset metadata defines two categories—"wildfire" (ID 0) and "fire" (ID 1)—our distribution analysis shows that 100% of the annotations (959/959) are tagged as "fire." The "wildfire" category exists only as a parent supercategory in the JSON structure and contains zero active labels.

Strategic Conclusion:
Given that we have a one-week development window, we have decided to simplify the project into a single-class detection task. Since "fire" is a sub-type of "wildfire," focusing exclusively on the "fire" label allows us to maximize model precision and avoid the technical debt of handling an empty class. This approach ensures a functional, high-performance model within the project's time constraints.
